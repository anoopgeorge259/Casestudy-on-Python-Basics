1.Bank Account Manager

a. Simulate a bank account with deposit, withdrawal, and balance enquiry operations

b. Use variables and data types to store account information

c. Implement conditional statements to check for sufficient balance before withdrawal

d. Use functions to perform deposit, withdrawal, and balance inquiry operations

```python
account_holder = "Anoop"
balance = 10000

def deposit(amount):
    global balance
    balance += amount
    print("Deposit successful")
    print("Updated balance:", balance)

def withdraw(amount):
    global balance
    if amount <= balance:
        balance -= amount
        print("Withdrawal successful")
        print("Remaining balance:", balance)
    else:
        print("Insufficient balance")

def check_balance():
    print("Account Holder:", account_holder)
    print("Current Balance:", balance)

while True:
    print("\nBank Account Manager")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = float(input("Enter deposit amount: "))
        deposit(amount)

    elif choice == 2:
        amount = float(input("Enter withdrawal amount: "))
        withdraw(amount)

    elif choice == 3:
        check_balance()

    elif choice == 4:
        print("Thank you for using the Bank System")
        break

    else:
        print("Invalid choice")
```

2.Weather Data Analyzer

a. Analyze weather data (temperature, humidity, etc.) for a given region

b. Use variables and data types to store weather data

c. Implement conditional statements to determine weather conditions (sunny, rainy, etc.)

d. Use functions to calculate average temperature and humidity, and print results

```python
def average_temperature(temp_list):
    return sum(temp_list) / len(temp_list)

def average_humidity(hum_list):
    return sum(hum_list) / len(hum_list)

def weather_condition(temp, hum):
    if temp > 32:
        return "Sunny"
    elif hum > 70:
        return "Rainy"
    else:
        return "Cloudy"

region = input("Enter the region: ")
days = int(input("Enter number of days of weather data: "))
temperature_list = []
humidity_list = []

for i in range(days):
    print("\nDay", i + 1)

    temp = float(input("Enter temperature: "))
    hum = float(input("Enter humidity: "))

    temperature_list.append(temp)
    humidity_list.append(hum)

print("\n------ Weather Report ------")
print("Region:", region)
print("\nDaily Weather Conditions:")

for i in range(days):
    condition = weather_condition(temperature_list[i], humidity_list[i])
    print("Day", i+1, ":", condition)

avg_temp = average_temperature(temperature_list)
avg_hum = average_humidity(humidity_list)
print("\nAverage Temperature:", avg_temp)
print("Average Humidity:", avg_hum)
print("\nCurrent Weather Condition:",
      weather_condition(temperature_list[-1], humidity_list[-1]))
```
